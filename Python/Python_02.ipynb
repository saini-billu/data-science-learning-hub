# **Question 1:What are strings and lists in Python? Explain the main differences between them with suitable examples.**

**What is a String in Python?**

A string is a sequence of characters used to store text, such as names, words, sentences and symbols. Strings are written inside single quotes (' ') or double quotes (" "). Strings are immutable, which means their characters cannot be changed after creation.

```python
#Example:

name = "Python"

print(name)
print(type(name))
```

**What is a List in Python?**

A list is an ordered collection of elements used to store multiple values in a single variable. A list can contain different data types such as integers, strings and floats. Lists are mutable, which means we can add, remove or modify elements after creation. Lists are created using square brackets [ ].

```python
student = ["Rahul", 23, 85.5]

print(student)
print(type(student))
```

**Main Differences Between Strings and Lists**

**1. Meaning**

**String**: A string is a sequence of characters used to store text.


**List**: A list is an ordered collection of elements used to store multiple values.

**2. Data Storage**

**String**: A string stores characters, words and sentences.

**List**: A list can store different types of data, such as integers, strings and floats.

**3. Mutability**

**String**: Strings are immutable, which means their characters cannot be changed after creation.

**List**: Lists are mutable, which means their elements can be changed after creation.

**4. Syntax**

**String**: Strings are created using single quotes or double quotes.

Example: name = "Rahul"

**List**: Lists are created using square brackets [ ].

Example: numbers = [10, 20, 30]

**5. Modification**

**String**: We cannot directly modify individual characters of a string after creation.

**List**: We can directly modify individual elements of a list using their index.

**6. Duplicate Values**

**String**: Strings can contain repeated characters.

**List**: Lists can also contain duplicate elements.

**7. Operations**

**String**: Strings support operations such as concatenation, slicing and repetition.

**List**: Lists support operations such as adding, removing, updating and sorting elements.

**Practical Example: Difference Between String and List:**

# **Question 2: What are indexing and slicing in Python strings? Explain positive indexing, negative indexing, and slicing with examples.**

**What is Indexing in Python?**

Indexing is a process of accessing an individual character from a string using its position (index number).

In Python, indexing starts from 0, not 1. Each character in a string has a unique index position.

```python
#Example of Indexing:

name = "PYTHON"
print(name[0])
print(name[1])
print(name[2])
```

**There are two types of indexing in Python:**

**A. Positive Indexing:**

Definition: Positive indexing is used to access characters in a string from left to right, starting with index 0.



```python
#Example:

text = "PYTHON"

print(text[0])
print(text[2])
print(text[5])
```

**B. Negative Indexing:**

Definition: Negative indexing is used to access characters in a string from right to left, starting with index -1.

```python
#Example:

text = "PYTHON"

print(text[-1])
print(text[-2])
print(text[-3])
```

**2. Slicing in Python Strings:**


Definition: Slicing is the process of extracting a specific portion or group of characters from a string using index positions.


Syntax:   string[start:stop:step]

```python
#Example:

text = "PYTHON"

print(text[0:3])
print(text[1:4])

# Negative Indexing

print(text[-4:-1])
print(text[-3:])

#Slicing with step

text = "PYTHON"

print(text[::2])
print(text[::-1])
```

# **Question 3: Explain string concatenation, repetition, escape sequences, and string formatting using format() and f-strings**

**1. String Concatenation:**

String concatenation means joining two or more strings together using the + operator.

```python
#Example:
first_name = "John"
last_name = "Doe"

full_name = first_name + " " + last_name
print(full_name)
```

**2. String Repetition:**

String repetition means repeating a string multiple times using the * operator.

```python
text = "Hello "
print(text * 3)
```

**3. Escape Sequences:**

Escape sequences are special characters used inside strings to perform specific formatting functions. They begin with a backslash (\).

scape Sequence:

 \n	  : New line

 \t	  : Tab space

 \\	  : Backslash

 \'	  : Single quote

 \"	  : Double quote

```python
print("Hello\nWorld") #New line

print("Name\tAge")    #Tab Space(\t)

print("He said, \"Python is easy\"")   #Using Quotes Inside String
```

**4. String Formatting using format():**


The format() method inserts values into placeholders {} within a string.

```python
name = "Alice"
age = 22

print("My name is {} and I am {} years old.".format(name, age))
```

**5. String Formatting using f-strings:**

f-strings (formatted string literals) were introduced in Python 3.6 and provide a simple and readable way to embed variables directly inside strings.


f"string {variable}"

```python
name = "Alice"
age = 22

print(f"My name is {name} and I am {age} years old.")
```

**Answer:**  


Python lists provide several built-in methods to manipulate data efficiently. Below are six commonly used list methods with examples:


**1. append()**

Purpose: Adds an element at the end of the list.

```python
numbers = [5, 2, 8, 1, 3]
numbers.sort()

print(numbers)
```

Purpose: Reverses the order of elements in the list.

Syntax:
       list.reverse()
   

```python
#Example:

numbers = [1, 2, 3, 4, 5]
numbers.reverse()

print(numbers)
```

#**Question 5: What is list comprehension in Python? Explain how it is different from creating a list using a normal loop. Also explain nested lists with an example**

**Answer. 5:**

**List Comprehension in Python**

List comprehension is a concise and efficient way to create lists in Python using a single line of code. It combines the creation of a list and the processing of its elements into one statement.

Syntax:

new_list = [expression for item in iterable if condition]

```python
squares = [x**2 for x in range(1, 6)]
print(squares)
```

**Creating a List Using a Normal Loop :**

The same task can be performed using a traditional for loop.

```python
#Example:

squares = []

for x in range(1, 6):
    squares.append(x**2)

print(squares)
```

**Difference Between List Comprehension and Normal Loop :**

**1. Code Length:**

List Comprehension: Short and concise code.

Normal Loop: More and longer code.

**2. Readability:**

**List Comprehension:** Easy to read for simple operations.

**Normal Loop:** Can be lengthy but easier for beginners to understand.

**3. Execution Speed:**

**List Comprehension**: Generally faster.

**Normal Loop:** Slightly slower due to multiple statements.

**4. Syntax:**

**List Comprehension:** Uses a single-line syntax.

**Normal Loop:** Uses multiple lines with for loop and append().

**5. Memory Usage:**

**List Comprehension:** More optimized in many cases.

**Normal Loop:** May use slightly more resources.

**6. Best Use Case:**

**List Comprehension:** Best for simple list creation and transformations.

**Normal Loop:** Best for complex logic and multiple operations.


```python
# Example : Using List Comprehension:

squares = [x**2 for x in range(1, 6)]
print(squares)
```

```python
#Using Normal Loop:
squares = []

for x in range(1, 6):
    squares.append(x**2)

print(squares)
```

##**Question 6: Write a Python program that accepts a sentence from the user and displays:**● The sentence in uppercase. ● The sentence in lowercase. ● The total number of characters. ● The first five characters. ● The last five characters. ● The sentence in reverse order

```python
#Python Program: String Operations on a Sentence

# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Display required information

#upper() → Converts the sentence to uppercase.
print("Sentence in uppercase:", sentence.upper())

#lower() → Converts the sentence to lowercase.
print("Sentence in lowercase:", sentence.lower())

#len() → Returns the total number of characters
print("Total number of characters:", len(sentence))

#sentence[:5] → Extracts the first five characters
print("First five characters:", sentence[:5])

#sentence[-5:] → Extracts the last five characters
print("Last five characters:", sentence[-5:])

#sentence[::-1] → Reverses the sentence.
print("Sentence in reverse order:", sentence[::-1])
```

# **Question 7: Create the following list:**  

numbers = [45, 12, 78, 34, 56]

Write a Python program to:
1. Add 90 to the list.
2. Insert 25 at index 2.
3. Remove 34.
4. Sort the list.
5. Display the largest and smallest values.
6. Reverse the final list.
Reverse the final list..

# **Question 8: Consider the following string: **

**languages = "Python,Java,C++,JavaScript" **


**Write a Python program to:**

1. Convert the string into a list.
2. Display each language separately.
3. Add "SQL" to the list.
4. Sort the list alphabetically.
5. Join all languages using " | ".

```python
languages = "Python,Java,C++,JavaScript"

# 1. Convert the string into a list
lang_list = languages.split(",")

# 2. Display each language separately
print("Languages:")
for language in lang_list:
    print(language)

# 3. Add "SQL" to the list
lang_list.append("SQL")

# 4. Sort the list alphabetically
lang_list.sort()

# 5. Join all languages using " | "
result = " | ".join(lang_list)

print("\nSorted List:", lang_list)
print("Joined String:", result)
```

# **Question 9: Given the following list:**

numbers = [2, 5, 8, 11, 14, 17, 20]

**Write a Python program using list comprehension to:**

1. Create a new list containing only even numbers.
2. Create another list containing the square of every number.
3. Display both lists.


```python
numbers = [2, 5, 8, 11, 14, 17, 20]

# 1. Create a list containing only even numbers
even_numbers = [num for num in numbers if num % 2 == 0]

# 2. Create a list containing squares of all numbers
square_numbers = [num**2 for num in numbers]

# 3. Display both lists
print("Even Numbers:", even_numbers)
print("Square Numbers:", square_numbers)
```

**Question 10: Consider the following student data:**

**students = [
["Rahul", 78],
["Priya", 92],
["Anil", 65],
["Meena", 88]
]**


**Write a Python program to:**
1. Display the name and mark of each student.
2. Find the student with the highest mark.
3. Calculate the average mark.
4. Create a separate list containing only student names.

```python
students = [
    ["Rahul", 78],
    ["Priya", 92],
    ["Anil", 65],
    ["Meena", 88]
]

# 1. Display the name and mark of each student
print("Student Details:")
for student in students:
    print("Name:", student[0], ", Marks:", student[1])

# 2. Find the student with the highest mark
highest_student = max(students, key=lambda x: x[1])
print("\nStudent with Highest Mark:", highest_student[0])
print("Highest Mark:", highest_student[1])

# 3. Calculate the average mark
total_marks = sum(student[1] for student in students)
average_marks = total_marks / len(students)
print("\nAverage Marks:", average_marks)

# 4. Create a separate list containing only student names
student_names = [student[0] for student in students]
print("\nStudent Names:", student_names)
```
