Dynamic typing means that we do not need to specify the data type of a variable while creating it.

**2. Float (float):**

**3. String (str):**

# **Question 3: Explain the following types of operators in Python with suitable examples for Arithmetic operators, Comparison operators, Logical operators, Assignment operators, Membership operators, Identity operators. Also explain the difference between implicit and explicit type conversion.**

**Answer 3:**

**What are Operators in Python?**

Operators are special symbols or keywords used to perform different operations on variables and values.

**Types of Operators in Python:**

**1. Arithmetic Operators:**

Arithmetic operators are used to perform mathematical operations such as addition, subtraction, multiplication, division, etc.


```python
#Example
a = 10
b = 3

print(a + b)   #Addition
print(a - b)   #Substraction
print(a * b)   #Multipcation
print(a / b)   #Division
print(a % b)   #Modulus
print(a // b)  #Floor Division
print(a ** b)  #Exponent
```

**2. Comparison Operators:**

Comparison operators are used to compare two values.

They return a Boolean result: True or False.

Operator : ==,!=,>,<,>=

```python
a=49
b=34

print(a<b)
print(a>=b)
print(a<=b)
```

**3. Logical Operators:**

Logical operators are used to combine two or more conditions.

Python provides three logical operators:

*  **and** : True if both conditions are true

*  **or** : True if at least one condition is true

*  **not** : Reverses the result

```python
#Example:

a = 10
b = 5

print(a > 5 and b < 10)
print(a > 15 or b < 10)
print(not(a > 5))
```

**4. Assignment Operators:**

Assignment operators are used to assign values to variables and update their values.

Operators: =,+=,-=,*=,/=,%=,//=,**=

```python
#Example:
a = 10

a += 5
print(a)

a *= 2
print(a)

a -= 5
print(a)
```

**5. Membership Operators:**

Membership operators are used to check whether a value is present in a sequence or collection, such as a string, list, tuple or set.

Python provides two membership operators:
- in: Returns True if the value is present.
- not in: Returns True if the value is not present.

```python
#Example:

numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 in numbers)
print(50 not in numbers)
```

**6. Identity Operators:**

Identity operators are used to check whether two variables refer to the same object in memory.

Python provides two identity operators:
- is: Returns True if both variables refer to the same object.
- is not: Returns True if both variables refer to different objects.

```python
#Example:

a = [10, 20, 30]
b = a
c = [10, 20, 30]

print(a is b)
print(a is c)
print(a is not c)
```

**What is Type Conversion in Python?**

Type conversion means converting one data type into another data type.

For example, converting an integer into a float or a string into an integer.

There are two types of type conversion in Python:
1. Implicit Type Conversion
Implicit type conversion is performed automatically by Python without the programmer converting the data type manually.

```python
#Example:

a = 10
b = 5.5

c = a + b

print(c)
print(type(c))
```

**2. Explicit Type Conversion:**

Explicit type conversion is performed manually by the programmer using built-in conversion functions.

```python
a = "100"

b = int(a)

print(b)
print(type(b))
```

**Difference Between Implicit and Explicit Type Conversion :**

1) Performed :

Implicit Conversion: Done automatically by Python.
Explicit Conversion: Done manually by the programmer.

2) Also Known As :

Implicit Conversion: Type Coercion.
Explicit Conversion: Type Casting.

3) Control :

Implicit Conversion: No control by programmer.
Explicit Conversion: Full control by programmer.

4) Functions Required :

Implicit Conversion: No conversion function required.
Explicit Conversion: Uses functions like int(), float(), str(), etc.

5) Data Loss :

Implicit Conversion: Generally avoids data loss.
Explicit Conversion: May cause data loss during conversion.

6) Effort :

Implicit Conversion: Requires less coding effort.
Explicit Conversion: Requires manual coding effort.

7) Example :

Implicit Conversion: 10 + 2.5 = 12.5 (Python converts 10 to 10.0).
Explicit Conversion: int(2.5) = 2 (Programmer converts float to integer).

8) Purpose :

Implicit Conversion: Ensures smooth execution of operations.
Explicit Conversion: Converts data into the required type as needed.

# **Question 5: What is a list in Python? Explain why lists are called mutable. Describe any five commonly used list methods with examples**

**Answer. 5**

A list is an ordered collection of items in Python. It can store elements of different data types such as integers, strings, floats, etc. Lists are enclosed within square brackets [ ].

**Why are Lists called Mutable?**

A list is called mutable because its elements can be changed, added, or removed after the list is created.




```python
#Example:
numbers = [10, 20, 30]

numbers[1] = 50

print(numbers)
```

**Five Commonly Used List Methods:**

1. append()

Purpose: Adds an element at the end of the list.

```python
#Example:
fruits = ["Apple", "Banana"]
fruits.append("Mango")

print(fruits)
```

2. insert()

Purpose: Inserts an element at a specified position.

```python
#Example:

numbers = [10, 20, 30]
numbers.insert(1, 15)

print(numbers)
```

3. remove()

Purpose: Removes the first occurrence of a specified element.

```python
fruits = ["Apple", "Banana", "Mango"]
fruits.remove("Banana")

print(fruits)
```

4. pop()

Purpose: Removes and returns the element at a specified index (or the last element if no index is given).

```python
#Example:
numbers = [10, 20, 30]
numbers.pop()

print(numbers)
```

5. sort()

Purpose: Arranges elements in ascending order.

```python
#Example:

numbers = [40, 10, 30, 20]
numbers.sort()

print(numbers)
```

# **Question 7:Write a Python program that accepts a sentence from the user and displays:**
1. The sentence in uppercase.
2. The sentence in lowercase.
3. The total number of characters.
4. The first five characters.
5. The reversed sentence.
6. The number of words in the sentence.

```python
# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Display required outputs
print("\nResults:")
print("1. Uppercase:", sentence.upper())
print("2. Lowercase:", sentence.lower())
print("3. Total number of characters:", len(sentence))
print("4. First five characters:", sentence[:5])
print("5. Reversed sentence:", sentence[::-1])
print("6. Number of words:", len(sentence.split()))
```

# **Question 8: Create the following list:**
marks = [72, 85, 60, 91, 78]
Write a Python program to:
1. Add 88 to the list.
2. Remove 60.
3. Sort the list.
4. Display the highest mark.
5. Display the lowest mark.
6. Calculate the average mark.

```python
# Create the list
marks = [72, 85, 60, 91, 78]

# 1. Add 88 to the list
marks.append(88)

# 2. Remove 60
marks.remove(60)

# 3. Sort the list
marks.sort()

# 4. Display the highest mark
highest = max(marks)

# 5. Display the lowest mark
lowest = min(marks)

# 6. Calculate the average mark
average = sum(marks) / len(marks)

# Display results
print("Updated List:", marks)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Average Mark:", average)
```

# **Question 9: Consider the following string:**

"Python Java C C++ Python Java Python"

Write a Python program to:
1. Convert the string into a list of programming languages.
2. Display the list.
3. Count how many times "Python" appears.
4. Remove duplicate language names.
5. Join the unique language names using " | ".

```python
# Given string
languages = "Python Java C C++ Python Java Python"

# 1. Convert string into a list
lang_list = languages.split()

# 2. Display the list
print("List of Languages:", lang_list)

# 3. Count occurrences of Python
python_count = lang_list.count("Python")
print("Number of times Python appears:", python_count)

# 4. Remove duplicate language names
unique_languages = list(set(lang_list))
print("Unique Languages:", unique_languages)

# 5. Join unique language names using " | "
joined_languages = " | ".join(unique_languages)
print("Joined String:", joined_languages)
```

# **Question 10: Write a Python program to store the following student information:**

● Student name

● Course name

● Marks in three subjects

The program should:
1. Store the three marks in a list.
2. Calculate the total marks.
3. Calculate the average.
4. Display the highest mark.
5. Display all student details using an f-string.

```python
# Accept student details
student_name = input("Enter Student Name: ")
course_name = input("Enter Course Name: ")

# Input marks for three subjects
mark1 = float(input("Enter marks of Subject 1: "))
mark2 = float(input("Enter marks of Subject 2: "))
mark3 = float(input("Enter marks of Subject 3: "))

# Store marks in a list
marks = [mark1, mark2, mark3]

# Calculate total, average, and highest mark
total = sum(marks)
average = total / len(marks)
highest = max(marks)

# Display all details using f-string
print(f"""
Student Name : {student_name}
Course Name  : {course_name}
Marks        : {marks}
Total Marks  : {total}
Average Marks: {average}
Highest Mark : {highest}
""")
```
